```python

print("Hit Barvaliya")
print("24DCE010")

import pandas as pd
import json

with open('customer_churn.json') as f:
    data = json.load(f)

df = pd.json_normalize(data)
print(df)
```

```python
df.info()
```

```python

# Rename columns
df.rename(columns={
    'profile.age': 'age',
    'profile.city': 'city',
    'transactions.total_spent': 'total_spent',
    'transactions.num_orders': 'num_orders',
    'engagement.last_login_days': 'last_login_days',
    'engagement.avg_session_time': 'avg_session_time'
}, inplace=True)

numeric_columns = []

for col in df.columns:
    if df[col].dtype != 'object':   
        numeric_columns.append(col)


print("Numeric columns:", numeric_columns)
```

```python
# fill with 0
df['total_spent'] = df['total_spent'].fillna(0)

# fill age with median
df['age'] = df['age'].fillna(df['age'].median())


# fill with mean
df['avg_session_time'] = df['avg_session_time'].fillna(df['avg_session_time'].mean())

# this is one-hot encoding
if 'city' in df.columns:
    df = pd.get_dummies(df, columns=['city'])

# drop customer_id if it exists
if 'customer_id' in df.columns:
    df.drop(columns=['customer_id'], inplace=True)
```
