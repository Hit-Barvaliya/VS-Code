```python
# Task 3: Dataset Exploration Using Python
# Dataset: Iris

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add target column
df['species'] = iris.target

# Replace numeric values with species names
df['species'] = df['species'].map({
    0: 'setosa',
    1: 'versicolor',
    2: 'virginica'
})

# Display first 10 records
print("First 10 Records:")
print(df.head(10))

# Number of rows and columns
print("\nDataset Shape:")
print(df.shape)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# Missing Values

print("\nMissing Values:")
print(df.isnull().sum())

# Summary Statistics
print("\nSummary Statistics:")
print(df.describe())

# Bar Chart
plt.figure(figsize=(6,4))
df['species'].value_counts().plot(kind='bar')
plt.title("Number of Flowers in Each Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.grid(axis='y')
plt.show()

# Histogram
plt.figure(figsize=(8,5))
plt.hist(df['sepal length (cm)'], bins=10)
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
# Task 4: Data Cleaning Exercise
# (Continue from Task 3)

print("Current Dataset:")
print(df.head())

# Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate Records
print("\nNumber of Duplicate Records:")
print(df.duplicated().sum())

# Handle Missing Values

# Numerical Columns
numeric_columns = df.select_dtypes(include="number").columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

# Categorical Columns
categorical_columns = df.select_dtypes(include=["object", "category", "string"]).columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

# Remove Duplicate Records
df = df.drop_duplicates()

print("\nShape After Removing Duplicates:")
print(df.shape)

# Data Type Conversion
df["species"] = df["species"].astype("category")

print("\nData Types:")
print(df.dtypes)

# Verification
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Records After Cleaning:")
print(df.duplicated().sum())

# Save Cleaned Dataset
df.to_csv("Iris_Cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```

```python
# Task 5: Data Visualization Dashboard

# Import libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Pie Chart

plt.figure(figsize=(6,6))

df["species"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Species Distribution")
plt.ylabel("")
plt.show()

# Line Chart

plt.figure(figsize=(8,5))

plt.plot(df.index, df["sepal length (cm)"], marker="o")

plt.title("Sepal Length")
plt.xlabel("Record Number")
plt.ylabel("Sepal Length (cm)")
plt.grid(True)

plt.show()


# Scatter Plot

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="sepal length (cm)",
    y="petal length (cm)",
    hue="species",
    s=80
)

plt.title("Sepal Length vs Petal Length")
plt.show()


# Box Plot

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="species",
    y="petal width (cm)"
)

plt.title("Petal Width by Species")
plt.show()


# Conclusion

print("Task 5 Completed Successfully!")
print("Generated:")
print("1. Pie Chart")
print("2. Line Chart")
print("3. Scatter Plot")
print("4. Box Plot")
```
