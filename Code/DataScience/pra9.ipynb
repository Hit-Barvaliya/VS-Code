```python
import numpy as np
import pandas as pd
from scipy import stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd

np.random.seed(42)

# Generate synthetic data
region_A = np.random.normal(500, 50, 40)
region_B = np.random.normal(520, 55, 40)
region_C = np.random.normal(560, 60, 40)

df = pd.DataFrame({
    'Sales': np.concatenate([region_A, region_B, region_C]),
    'Region': ['A'] * 40 + ['B'] * 40 + ['C'] * 40
})

print(df.head())
df.info()
print(df.tail())
```

```python
# Assumption Checkuing
# Normality Test (Shapiro-Wilk Test)

print(stats.shapiro(region_A))
print(stats.shapiro(region_B))
print(stats.shapiro(region_C))
```

```python
# Equal Variance Test (Levene's Test)

print(stats.levene(region_A, region_B, region_C))

```

```python
# Independent t-test (Region A vs B)

t_stat, p_value = stats.ttest_ind(region_A, region_B)
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
#Paired t-test (Before vs After Strategy)

before = np.random.normal(500, 50, 30)
after = before + np.random.normal(20, 10, 30)

stats.ttest_rel(before, after)
```

```python
# One-Way ANOVA

model = ols('Sales ~ C(Region)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

```python
# Post-Hoc Test (Tukey HSD)

tukey = pairwise_tukeyhsd(endog=df['Sales'], groups=df['Region'], alpha=0.05)

print(tukey)
```
