```python
import json
import pandas as pd

json_data = """
{
 "name" : "Ahmedabad",
 "main" : {
  "temp" : 34.5,
  "humidity" : 62
 },
 "wind" : {
  "speed" : 5.1
 }
}
  """

data = json.loads(json_data)

json_table = {
    "City" : data["name"],
    "Temperature" : data["main"]["temp"],
    "Humidity" : data["main"]["humidity"],
    "wind_speed" : data["wind"]["speed"]
}

df_json = pd.DataFrame([json_table])
print("\nJSON + TABULAR FORMAT")
print(df_json)
```

```python
xml_data = """
<rss>
  <channel>
    <item>
      <title>Server Maintenance</title>
      <priority>High</priority>
    </item>
    <item>
      <title>System Upgrade</title>
      <priority>Medium</priority>
    </item>
  </channel>
</rss>
"""
root = ET.fromstring(xml_data)

records = []
for item in root.findall(".//item"):
  records.append({
    "title" : item.find("title").text,
    "priority" : item.find("priority").text
  })

df_xml = pd.DataFrame(records)
print("\nXML + TABULAR FORMAT")
print(df_xml)
```

```python
text = """
Dear Customer,
Your order has been successfully delivered on 12/09/2024!
Thank you for shopping with us.
Regards,
Support Team
"""

import pandas as pd
import re
import nltk
# nltk.download('stopwords')
from nltk.corpus import stopwords

from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer


# Lowercase
text = text.lower()

#remove punctuation & numbers
text =re.sub("[^a-zA-Z]", " ", text)

# Tokenization
tokens = text.split()

#stopword removal
stop_words = set(stopwords.words("english"))
tokens_clean = [w for w in tokens if w not in stop_words]

# Stemming vs Lemmatization
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed = [stemmer.stem(word) for word in tokens_clean]
lemmatized = [lemmatizer.lemmatize(word) for word in tokens_clean]

print("Original Tokens (Cleaned):", tokens_clean)
print("Stemmed Tokens:", stemmed)
print("Lemmatized Tokens:", lemmatized)

documents =[" ".join(tokens_clean)]

vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(documents)

df_bow = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
    )
print("\nBAG OF WORDS")
print(df_bow)
```
