```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_excel("C:\\Users\\Dell\\OneDrive\\Desktop\\Book1.xlsx")

sns.histplot(df['SpealLengthCm'], kde=True)
plt.title('Distribution of SpealLengthCm')
plt.xlabel('SpealLengthCm')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.boxplot(df['SpealLengthCm'])
```

```python
# this is for Min Max normilization

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df_minmax = df.copy()
print(df_minmax)
df_minmax[['SpealLengthCm','SpealWidthCm','PetalLengthCm','PetalWidthCm']] = scaler.fit_transform(
    df[['SpealLengthCm','SpealWidthCm','PetalLengthCm','PetalWidthCm']]
)
print(df_minmax)
```

```python
# this is z score normuilization

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_zscore = df.copy()
print(df_zscore)
df_zscore[['SpealLengthCm','SpealWidthCm','PetalLengthCm','PetalWidthCm']] = scaler.fit_transform(
    df[['SpealLengthCm','SpealWidthCm','PetalLengthCm','PetalWidthCm']]
)
print(df_zscore)
```

```python
# this is for decimal Scaling

df_decimal = df.copy()
print(df_decimal)
df_decimal[['SpealLengthCm','SpealWidthCm','PetalLengthCm','PetalWidthCm']] = \
df[['SpealLengthCm','SpealWidthCm','PetalLengthCm','PetalWidthCm']] / 10
print(df_decimal)
```

```python
# label encoding

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df2 = df.copy()

df2['Species_Label'] = le.fit_transform(df['Species'].astype(str))

print(df2)
```

```python

print(df)
df_onehot = pd.get_dummies(df, columns=['Species'])
print(df_onehot)
```

```python
# This is binning

df['PetalLength_bin'] = pd.cut(df['PetalLengthCm'], bins=3, labels=['Small','Medium','Large'])
print(df)
```

```python
# this is equal frequency binning

df['PetalLength_bin2'] = pd.qcut(df['PetalLengthCm'], q=3, labels=['Low','Medium','High'])
print(df)
```

```python
# This is custom binning

bins = [0, 3, 6, 10]
labels = ['Small', 'Medium', 'Large']

df['PetalLength_bin3'] = pd.cut(df['PetalLengthCm'], bins=bins, labels=labels)
print(df)
```
